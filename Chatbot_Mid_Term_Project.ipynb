```python
# Simple chatbot program

precepts = [
    "hello",
    "laptop",
    "school",
    "none",
    "zoom",
    "$1000",
    "storage",
    "colors",
    "yes",
    "no"
]
# Define a list of known questions and their responses

responses = [
    "Hello! Can I get your name?",
    "I'd love to help you with that! Are you looking to buy a laptop for school, or to help with work?",
    "Alright. Does your school have any required specifications or criteria that the laptop has to meet? If yes, please elaborate.",
    "Alright. Can you tell me what applications do you anticipate using on a daily basis?",
    "I've found some options for you. To narrow down this search further, can you tell me what's your budget?",
    "Take a look at Apple's Macbook Neo. It comes with 8GB of RAM, and a 13-inch display.",
    "It has two storage options-256GB and 512GB. I'd recommend you get the 256GB storage option for now. You can always upgrade later if you need to.",
    "It's available in four colors-silver, blush, citrus, and indigo.",
    "Your order is confirmed! You should be receiving more details via email soon to complete your purchase. Is there anything else I can help you with?",
    "Enjoy the rest of your day! We hope to serve you again."
]
```

```python
# Function to get user input and respond
def chatbot():

    # User welcome and termination instruction.
    print("Welcome! Enter 'bye' to exit.")

    # Main loop:
    while True:
        
        # 1. Get user input
        userInput = input("You: ")
                
        # 2. Check if the user wants to exit
        if userInput == "bye":
            print("Bot: Goodbye! Feel free to come back anytime.")
            break;
         
        # 3. Search for a pattern in the precepts. And fetch a predefined responses
        #   3.1 Check if any of precepts are there in the input
        #   3.2 Return the index
        #   3.3 Use index to find repsonse
        #   3.4 Break the loop
        responseFound = False
        for pattern in precepts:
            if pattern in userInput:
                responseFound = True
                index = precepts.index(pattern)
                response = responses[index]
                print("Bot: ", response)
                break;
                
        if responseFound == False:
            print("Bot: I'm sorry, I couldn't understand your request. Could you please phrase that in a different way?")
        
# Run the chatbot
# chatbot()
```

```python
# Write a function that takes in precept categories and responses lists and returns a mapping between them.

# - Use ZIP function to combine two independent lists (precepts and responses) and create a new map between them.
# - Search on the ZIP function and how to use it in a for loop
# - Run the main loop and check that this function works

# ___ Write your code here _____

def zip_responses(precepts, response):

    mapping = {}
    
    for key, val in zip(precepts, response):
        mapping[key] = [val]
    
    return mapping
```

### 2.3 Function : Detect pattern from input

---

Use regular expressions + dictionary from before to detect a pattern from user input. 

**Function input** : User Input ("Hi, how are you?")  <br>
**Output**         : Response_type ("Greetings") 

---

```python
## Function name: "detect_pattern"

## Create the detect_pattern function with one argument : user_input
## 1. Loop through each key in the dictionary:
## 2. Write if-else conditions to check for pattern match
##    2.1 For the if condition, use the re.search to check if the pattern exists in user input 
##    2.2 If there is a match, return the response type (e.g., "greetings" or "bye" or "happy")
##    2.3 If none of the conditions match, return "default" which will be the default response of the chatbot
##    Note: use else for last conditon

## TIP: To avoid multiple conditions from matching, use a break statement. 

# Give the function a characteristic name (e.g., detect_pattern) 
# ___ Write your code here _____

user_state = {
    "name": "",
    "color": ""
}

def detect_pattern(user_input):
    global user_state
    
    if "name is" in user_input.lower():
        name_match = re.search(r"name is\s+([a-zA-Z]+)", user_input, re.IGNORECASE)
        if name_match:
            user_state["name"] = name_match.group(1).capitalize()
            return "name_response"
            
    elif "i am" in user_input.lower():
        name_match = re.search(r"i am\s+([a-zA-Z]+)", user_input, re.IGNORECASE)
        if name_match:
            user_state["name"] = name_match.group(1).capitalize()
            return "name_response"
            
    color_match = re.search(r"(?:color is|i like(?:\s+the\s+color)?)\s+([a-zA-Z]+)|\b(silver|blush|citrus|indigo)\b", user_input, re.IGNORECASE)
    if color_match:
        detected_color = color_match.group(1) if color_match.group(1) else color_match.group(2)
        user_state["color"] = detected_color.lower()
        return "color_response"

    for pattern, category in re_patterns.items():
        if re.search(pattern, user_input, re.IGNORECASE):
            return category
            
    if user_state["name"] == "":
        name_match = re.search(r"^\s*([a-zA-Z]+)\s*$", user_input)
        if name_match:
            user_state["name"] = name_match.group(1).capitalize()
            return "name_response"
            
    return "default"
```

### 2.4 Function : Retrieve response

---

Use the output of pattern detection function (above) to trigger a chatbot response. 

In the function, use the "responses" dictionary from above. 


Follow the steps to write function. The function returns a response (e.g., "hello", or "how are you").

**Function input** : User Input ("Hi, how are you?")  <br>
**Detect pattern** : Response_type ("Greetings")  <br>
**Output**         : One of ["hello", "how are you?", etc...]

---

```python
## Function name: "chatbot_response" 

# Write a function to return a responses by indexing dictionary with response categories.

## Create the chatbot_response function with one argument : user_input
## 1. Call the detect_pattern function and store the response in variable (e.g., detected_type)
## 2. Write if-else conditions to check for pattern match
##    2.1 For the if condition, check if the detected_type is one of the keys of the dictionary
##    2.2 If there is a match, return the response from zipped responses
##    2.3 If none of the conditions, use "default" response

# Give the function a characteristic name (e.g., detect_pattern) 
# ___ Write your code here _____

def chatbot_response(user_input):
    detected_type = detect_pattern(user_input)
    
    name = user_state["name"]
    color = user_state["color"]
    
    if detected_type == "name_response":
        response = responses["name_response"][0]
        return response.format(name=name)
        
    elif detected_type == "color_response":
        response = responses["color_response"][0]
        return response.format(color=color)
        
    elif detected_type in responses:
        return responses[detected_type][0]
        
    else:
        return "I'm sorry, I couldn't understand your request. Could you please phrase that in a different way?"
```

```python
## A reminder for what the main loop looked like.
def chat():
    print("Chatbot: Welcome!")
    while True:
        user_input = input("You: ")

        # Generate chatbot response
        bot_response = chatbot_response(user_input)
        print(f"Chatbot: {bot_response}")

# Start chatting with the bot
# chat()
```

```python
# ___ Write your code here _____

import json
import re
import random
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('wordnet', quiet=True)

lemmatizer = WordNetLemmatizer()

user_memory = {
    "name": "Customer",
    "item": "order",
    "size": "medium",
    "side": "fries",
    "drink": "soda",
    "quantity": "1"
}

DISPLAY_NAMES = {
    "big mac": "Big Mac",
    "quarter pounder": "Quarter Pounder",
    "mccrispy": "McCrispy",
    "mcnuggets": "McNuggets",
    "mcchicken": "McChicken"
}

def load_files(filepath="intents.json"):
    try:
        with open(filepath, "r") as f:
            data = json.load(f)
    except FileNotFoundError:
        data = {
            "intents": [
                {"tag": "greeting", "patterns": ["hi", "hello"], "responses": ["Would you like to place an order today?"]},
                {"tag": "order_intent", "patterns": ["yes", "ready to order"], "responses": ["Awesome, {name}! What main item can I get you?"]}
            ],
            "synonyms": {}
        }
    
    rgx2int = {}
    int2res = {}
    
    for intent in data.get("intents", []):
        tag = intent["tag"]
        patterns_list = intent["patterns"]
        regex_key = "|".join([rf"\b({p})\b" for p in patterns_list])
        rgx2int[regex_key] = tag
        int2res[tag] = intent["responses"]
        
    synonym_rules = {}
    for base_word, synonym_list in data.get("synonyms", {}).items():
        for synonym in synonym_list:
            clean_syn = synonym.lower().replace("'", "")
            synonym_rules[clean_syn] = base_word.lower()
        
    return rgx2int, int2res, synonym_rules
```

```python
def preprocess_input(user_input, synonym_rules):
    clean_text = user_input.lower().replace("'", "")
    
    if synonym_rules:
        sorted_synonyms = sorted(synonym_rules.keys(), key=len, reverse=True)
        
        synonym_regex = re.compile(r'\b(' + '|'.join(map(re.escape, sorted_synonyms)) + r')\b')
        
        clean_text = synonym_regex.sub(lambda match: synonym_rules[match.group(0)], clean_text)
            
    clean_text = re.sub(r'[^\w\s]', '', clean_text)
    
    tokens = word_tokenize(clean_text)
    
    lemmatized_tokens = [lemmatizer.lemmatize(word, pos='v') for word in tokens]
            
    return " ".join(lemmatized_tokens)
```

```python
def extract_named_entities(user_input):
    global user_memory
    lowered = user_input.lower().replace("'", "")
    
    name_match = re.search(r"\b(?:name is|i am|call me)\s+([a-zA-Z]+)", user_input, re.IGNORECASE)
    if name_match:
        user_memory["name"] = name_match.group(1).capitalize()
    elif re.match(r"^\s*([a-zA-Z]+)\s*$", user_input) and user_memory["name"] == "Customer":
        user_memory["name"] = user_input.strip().capitalize()
        
    qty_match = re.search(r"\b(\d+|one|two|three|four|five)\s+(?:big mac|burger|mccrispy|mcnuggets|meal|combo|item)s?\b", lowered)
    if qty_match:
        user_memory["quantity"] = qty_match.group(1)

    item_match = re.search(r"\b(big mac|quarter pounder|mccrispy|mcnuggets|burger|mcchicken)\b", lowered)
    if item_match:
        matched_item_raw = item_match.group(1)
        user_memory["item"] = DISPLAY_NAMES.get(matched_item_raw, matched_item_raw.title())
        
    size_match = re.search(r"\b(small|medium|large|big|huge|max|supersize)\b", lowered)
    if size_match:
        size_word = size_match.group(1)
        if size_word in ["big", "huge", "max", "supersize"]: size_word = "large"
        user_memory["size"] = size_word
        
    side_match = re.search(r"\b(fries|french fries|chips|apple slices|salad|side salad)\b", lowered)
    if side_match:
        user_memory["side"] = side_match.group(1)
```

```python
def match_intent(processed_text, rgx2int):
    for pattern, tag in rgx2int.items():
        if re.search(pattern, processed_text, re.IGNORECASE):
            return tag
    return None
```

```python
def generate_response(user_input, rgx2int, int2res, synonym_rules):
    extract_named_entities(user_input)
    
    processed_input = preprocess_input(user_input, synonym_rules)
    
    matched_tag = match_intent(processed_input, rgx2int)
    
    if matched_tag and matched_tag in int2res:
        responses_list = int2res[matched_tag]
        selected_template = random.choice(responses_list)
        
        try:
            return selected_template.format(**user_memory)
        except KeyError:
            return selected_template
            
    return "I'm sorry, I couldn't understand your request. Could you please phrase that in a different way?"
```

```python
def chatbot():
    global user_memory
    user_memory = {"name": "Customer", "item": "order", "size": "medium", "side": "fries", "drink": "soda", "quantity": "1"}
    
    rgx2int, int2res, synonym_rules = load_files()
    print("=======================================================")
    print("🤖 McDonald's Automated Drive-Thru Assistant Activated")
    print("   Type 'exit' or 'quit' anytime to complete your trip. ")
    print("=======================================================")
    print("Bot: Would you like to place an order today?")

    while True:
        user_input = input("You: ").strip()
                
        if user_input.lower() in ["exit", "quit"]:
            print(f"Bot: Thank you for visiting, {user_memory['name']}! Enjoy your {user_memory['quantity']} {user_memory['item']} combo!")
            break
        
        if not user_input:
            continue
        
        reply = generate_response(user_input, rgx2int, int2res, synonym_rules)
        print(f"Bot: {reply}")
```

```python
def run_diagnostic_tests():
    print("--- Starting Automated Diagnostic Verification Framework ---")
    rgx2int, int2res, synonym_rules = load_files()
    
    test_str = "let's go order!"
    processed = preprocess_input(test_str, synonym_rules)
    print(f"Test 1 (Apostrophe Normalization): '{test_str}' -> '{processed}'")
    assert "yes" in processed or "order" in processed, "Failed Contraction Normalization handling."
    
    phrase_str = "i want a double cheeseburger combo"
    processed_phrase = preprocess_input(phrase_str, synonym_rules)
    print(f"Test 2 (Multi-Word Phrase Mapping): '{phrase_str}' -> '{processed_phrase}'")
    assert "big mac" in processed_phrase, "Failed Multi-Word Synonym Phrase Mapping replacement."
    assert "large mac" not in processed_phrase, "Corruption Bug Present: Substituted text was incorrectly mutated!"
    
    extract_named_entities("I want to buy 3 McCrispy sandwiches please")
    print(f"Test 3 (Regex Quantifier Capture): Captured Qty='{user_memory['quantity']}', Item='{user_memory['item']}'")
    assert user_memory["quantity"] == "3", "Advanced Regex Quantifier Failure."
    assert user_memory["item"] == "McCrispy", "Regex Item Extraction Failure."
    
    print("--- All Automated Test Suites Completed Successfully! [PASSED] ---\n")

run_diagnostic_tests()
```
